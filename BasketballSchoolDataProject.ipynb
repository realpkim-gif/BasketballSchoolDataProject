```python
import pandas as pd
import altair as alt

url = "https://raw.githubusercontent.com/sawula/Survey-of-Python-data-science-stack/refs/heads/gh-pages/NBA%20Stats%202324.csv"

nba = pd.read_csv(url)

average_ppg = nba['PPG'].mean()


alt.Chart(nba).mark_circle().encode(
    x='MPG',
    y='PPG',
    color="TEAM",
    tooltip=['POS',"MPG",'PPG',"TEAM","NAME"]
).properties(
    title='As minutes per game increases, PPG increases'
)
```

```python
import numpy as np
import matplotlib.pyplot as plt
#PREDICTS PPG BASED ON MPG if a player is on nba
np.set_printoptions(precision=3)

x = nba['MPG']
y = nba['PPG']

a, b = np.linalg.lstsq(np.column_stack((x**2, x)), y, rcond=None)[0]

def predict_ppg(mpg):
    return a * mpg**2 + b * mpg

ask = input("How many MPG do you have? ")
mpg_input = float(ask)
print(f"Predicted PPG: {predict_ppg(mpg_input):.3f}")
print(a,b)
mpg_range = np.linspace(0, float(ask)+0.5*mpg_input, 10000)
ppg_curve = predict_ppg(mpg_range)

plt.plot(mpg_range, ppg_curve, color='red', label='Quadratic Fit (Through Origin)')

plt.scatter(mpg_input, predict_ppg(mpg_input), c='blue', s=100, label=f'Your MPG: {mpg_input:.3f},\nYour PPG {predict_ppg(mpg_input):.3f}')

plt.xlabel('MPG')
plt.ylabel('PPG')
plt.title('NBA PPG vs MPG Quadratic Fit (expected PPG based on MPG)')
plt.legend()
plt.grid(True)
plt.show()
```

We learned at as MPG increases, PPG increases with a quadratic relationship.
